```python
# ============================================================
# INTELLIGENT LANGUAGE DETECTION SYSTEM
# Cell 2 - Language Detection Engine
# ============================================================

from langdetect import detect, detect_langs
import pandas as pd

# Language code → readable language name
LANGUAGE_NAMES = {
    "en": "English",
    "ta": "Tamil",
    "hi": "Hindi",
    "te": "Telugu",
    "ml": "Malayalam",
    "kn": "Kannada",
    "bn": "Bengali",
    "mr": "Marathi",
    "gu": "Gujarati",
    "pa": "Punjabi",
    "ur": "Urdu",
    "fr": "French",
    "de": "German",
    "es": "Spanish",
    "it": "Italian",
    "pt": "Portuguese",
    "ru": "Russian",
    "ja": "Japanese",
    "ko": "Korean",
    "zh-cn": "Chinese",
    "ar": "Arabic",
    "tr": "Turkish",
    "nl": "Dutch"
}

def detect_language(text):
    """
    Detect the primary language and confidence
    from the provided text.
    """

    if not text or not text.strip():
        return {
            "language": "No Input",
            "code": "—",
            "confidence": 0.0
        }

    try:
        text = text.strip()

        # Detect language probabilities
        predictions = detect_langs(text)

        best = predictions[0]

        code = best.lang
        confidence = best.prob * 100

        language = LANGUAGE_NAMES.get(
            code,
            code.upper()
        )

        return {
            "language": language,
            "code": code,
            "confidence": confidence
        }

    except Exception as error:
        return {
            "language": "Unable to Detect",
            "code": "—",
            "confidence": 0.0
        }


# Test
test_text = "This system can automatically identify the language of a sentence."

result = detect_language(test_text)

print("🧪 Test Result")
print("--------------------------------")
print("Language   :", result["language"])
print("Code       :", result["code"])
print("Confidence :", f"{result['confidence']:.2f}%")
```
